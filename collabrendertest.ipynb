```python
# @title files
#set the paths to your Blender Files

project_folder = "malia" #@param {type:"string"}
renderfile = "malia" #@param {type:"string"}
outputname = "malia" #@param {type:"string"}
!mkdir -p /content/drive/MyDrive/Blender/render_results/{folder}/shots



filename = f'/content/drive/MyDrive/Blender/{project_folder}/{renderfile}.blend'
outputfolder = f'/content/drive/MyDrive/blenderfiles/render_results/{project_folder}/shots/{outputname}'
```

```python
# @title render settings
# @title
# prompt: prompt: script to render the blender file from frame 10 to 20 in jpeg format using cuda resolution 4096 by 4096 colour high contrast a sample of 156 and camera focus of 20mm
frametorender = 41 #@param {type:"integer"}
endframe = frametorender
x = frametorender
y = endframe
resolution_x = 1920 #@param {type:"integer"}
resolution_y = 1920 #@param {type:"integer"}
fps = 24 #@param {type:"integer"}
samples = 4096 #@param {type:"integer"}
output_format = "PNG" #@param ["JPEG", "PNG"]
command2 = f'--python-expr "import bpy; bpy.context.scene.render.resolution_x = {resolution_x}; bpy.context.scene.render.resolution_y = {resolution_y}"'
command = f'--python-expr "import bpy ; bpy.data.scenes[\\"Scene\\"].cycles.samples ={samples}"'
command3 = f'--python-expr "import bpy; bpy.context.scene.render.fps = {fps}"'
command5 = f'--python-expr "import bpy; bpy.context.scene.render.image_settings.file_format = \'{output_format}\'"'

!./blender-{blender_version}-linux-x64/blender $filename -b -noaudio -E 'CYCLES' {command} {command2} {command3} {command5} -o $outputfolder -s {x} -e {y} -a -- --cycles-device CUDA
#!./blender-{blender_version}-linux-x64/blender -b {filename} -E CYCLES {command} -o {outputfolder} -s {x} -e {y} -a -- --cycles-device CUDA -- --render-output format=JPEG
#--python-expr "import bpy ; bpy.data.scenes[\"Scene\"].cycles.samples = VALUE"
#--python-expr "import bpy ; bpy.data.scenes[\"Scene\"].cycles.samples =10"
#!./blender-{blender_version}-linux-x64/blender -b {filename} -E CYCLES -o {outputfolder} -s {x} -e {y} -a -- --cycles-device CUDA -- --render-output format=JPEG resolution_x={resolutionx} resolution_y={resolutiony} color_management view_transform=High Contrast --samples {samples} --fps {fps}
```

```python
# @title files
#set the paths to your Blender Files

project_folder = "shine" #@param {type:"string"}
renderfile = "shine" #@param {type:"string"}
outputname = "shine" #@param {type:"string"}
!mkdir -p /content/drive/MyDrive/Blender/render_results/{folder}/animation

filename = f'/content/drive/MyDrive/Blender/{project_folder}/{renderfile}.blend'
outputfolder = f'/content/drive/MyDrive/Blender/{project_folder}/animation/{outputname}'
```

```python
# @title render settings
# prompt: script to render the blender file from frame 10 to 20 in jpeg format using cuda resolution 4096 by 4096 with frames per secnd30

startframe = 30 #@param {type:"integer"}
endframe = 80 #@param {type:"integer"}
x = startframe
y = endframe
resolution_x = 1080 #@param {type:"integer"}
resolution_y = 1920 #@param {type:"integer"}
samples = 320 #@param {type:"integer"}
fps = 24 #@param {type:"integer"}
output_format = "JPEG" #@param ["JPEG", "PNG"]

command2 = f'--python-expr "import bpy; bpy.context.scene.render.resolution_x = {resolution_x}; bpy.context.scene.render.resolution_y = {resolution_y}"'
command = f'--python-expr "import bpy ; bpy.data.scenes[\\"Scene\\"].cycles.samples ={samples}"'
command3 = f'--python-expr "import bpy; bpy.context.scene.render.fps = {fps}"'
command5 = f'--python-expr "import bpy; bpy.context.scene.render.image_settings.file_format = \'{output_format}\'"'


!./blender-{blender_version}-linux-x64/blender $filename -b -noaudio -E 'CYCLES' {command} {command2} {command3} {command5} -o $outputfolder -s {x} -e {y} -a -- --cycles-device CUDA
```
